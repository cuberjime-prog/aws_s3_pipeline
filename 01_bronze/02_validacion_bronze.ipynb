```python
%sql
---   primeros 5 registros
SELECT *
FROM aws_s3.ventas.bronze_ventas_raw
LIMIT 5;
```

```python
%sql
---  número de registros:
SELECT COUNT(*) AS total_registros
FROM aws_s3.ventas.bronze_ventas_raw;
```

```python
%sql
--- origen de los datos
SELECT
    source_file,
    COUNT(*) AS registros
FROM aws_s3.ventas.bronze_ventas_raw
GROUP BY source_file;
```
