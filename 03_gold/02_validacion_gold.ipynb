```python
%sql
--- Resumen general de ventas
SELECT
    SUM(cantidad_ventas) AS total_ventas,
    SUM(unidades_vendidas) AS total_unidades,
    ROUND(SUM(ingresos_crc), 2) AS ingresos_totales_crc,
    ROUND(AVG(ticket_promedio_crc), 2) AS ticket_promedio_diario_crc
FROM aws_s3.ventas.gold_ventas_diarias;
```

```python
%sql
--- Top 10 productos por ingresos
SELECT
    producto,
    cantidad_ventas,
    unidades_vendidas,
    ingresos_crc
FROM aws_s3.ventas.gold_ventas_producto
ORDER BY ingresos_crc DESC
LIMIT 10;
```

```python
%sql
--- Comparar los totales entre Silver y Gold
SELECT
    (SELECT COUNT(*)
     FROM aws_s3.ventas.silver_ventas) AS registros_silver,

    (SELECT COUNT(*)
     FROM aws_s3.ventas.gold_ventas_diarias) AS dias_con_ventas,

    (SELECT ROUND(SUM(venta_calculada_crc), 2)
     FROM (
         SELECT ROUND(
             cantidad_limpia * precio_unitario_crc_limpio
             * (1 - descuento_limpio / 100), 2
         ) AS venta_calculada_crc
         FROM aws_s3.ventas.silver_ventas
     )) AS ingresos_calculados_silver,

    (SELECT ROUND(SUM(ingresos_crc), 2)
     FROM aws_s3.ventas.gold_ventas_diarias) AS ingresos_gold;
```
