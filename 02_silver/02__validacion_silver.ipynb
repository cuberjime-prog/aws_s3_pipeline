```python
%sql
SELECT 'Bronze' AS capa, COUNT(*) AS registros
FROM aws_s3.ventas.bronze_ventas_raw

UNION ALL

SELECT 'Silver válidas', COUNT(*)
FROM aws_s3.ventas.silver_ventas

UNION ALL

SELECT 'Silver rechazadas', COUNT(*)
FROM aws_s3.ventas.silver_ventas_rechazadas;
```

```python
%sql
--- Revisar las ventas con totales incorrectos
SELECT
    id_venta,
    cantidad_limpia,
    precio_unitario_crc_limpio,
    descuento_limpio,
    total_crc_limpio,
    total_esperado
FROM aws_s3.ventas.silver_ventas
WHERE total_revisar = TRUE;
```

```python
%sql
--- Confirmar que no existan duplicados en las ventas válidas
SELECT
    id_venta,
    COUNT(*) AS cantidad
FROM aws_s3.ventas.silver_ventas
GROUP BY id_venta
HAVING COUNT(*) > 1;
```
